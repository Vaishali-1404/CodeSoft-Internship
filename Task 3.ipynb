```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("creditcard.csv")
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df['Class'].value_counts()
```

```python
sns.countplot( x = 'Class', data = df)
plt.title("Class Distribution ( 0 = Genuine, 1 = Fraud)")
plt.show()
```

```python
from sklearn.preprocessing import StandardScaler

SK = StandardScaler()
df['Amount'] = SK.fit_transform(df[['Amount']])
```

```python
df.columns
```

```python
from sklearn.model_selection import train_test_split

x = df.drop(['Class'],axis = 1)
y = df['Class']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 42) 
```

```python
from imblearn.over_sampling import SMOTE

sm = SMOTE(random_state = 42)
x_res, y_res = sm.fit_resample(x,y)
```

```python
from sklearn.linear_model import LogisticRegression

model_1 = LogisticRegression(max_iter = 5000)
model_1.fit(x_train, y_train)
```

```python
y_pred_1 = model_1.predict(x_test)
print(y_pred_1)
```

```python
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score

# Model 1
accuracy = accuracy_score(y_test, y_pred_1)
print("Accuracy : ", accuracy)

print("Classification Report :\n", classification_report(y_test, y_pred_1))
```

```python
cm = confusion_matrix(y_test, y_pred_1)
sns.heatmap(cm, annot = True, cmap = 'Reds')
plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

```python
from sklearn.ensemble import RandomForestClassifier

model_2 = RandomForestClassifier(n_estimators = 100, random_state = 42)
model_2.fit(x_train, y_train)
```

```python
y_pred_2 = model_2.predict(x_test)
print(y_pred_2)
```

```python
# Model 2
print("Accuracy:", accuracy_score(y_test, y_pred_2))

print("Classification Report:\n", classification_report(y_test, y_pred_2))
```

```python
cm = confusion_matrix(y_test, y_pred_2)

sns.heatmap(cm, annot = True, cmap = 'Greens')
plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```
